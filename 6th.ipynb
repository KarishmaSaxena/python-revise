    Literals refers to the fixed values that are assigned to a variable. For example, in the statement a = 5, the value 5 is a literal.

types of literals in python are as follows:
1. String literals
2. Numeric literals
3. Boolean literals
4. Special literals

string literals are used to represent text data. They can be enclosed in single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """). For example:

special literals are used to represent special values in Python. The most common special literal is None, which represents the absence of a value or a null value. 

```python
single_quote_string = 'Hello, World!'
double_quote_string = "Python is awesome!"
triple_quote_string = '''This is a multi-line string.'''
print(single_quote_string, double_quote_string, triple_quote_string, triple_quote_string, sep='\n')
```

numeric literals are used to represent numbers. They can be integers, floating-point numbers, or complex numbers. For example:


```python
integer_literal = 42
floating_point_literal = 3.14
binary_literal = 0b1010
octal_literal = 0o52
hexadecimal_literal = 0x2A
Complex_literal = 2 + 3j
boolean = True + 4
b = False +10 
FLOAT_LITERAL = 1.23e-4
print(integer_literal, floating_point_literal, binary_literal, octal_literal, hexadecimal_literal, Complex_literal, FLOAT_LITERAL, boolean, b, sep='\n')
print(type(binary_literal))
print(type(octal_literal))
print(type(hexadecimal_literal))
print(type(Complex_literal))
print(type(FLOAT_LITERAL))
```

boolean literals represent the truth values True and False. They are used in conditional statements and logical operations. For example:


```python
is_true = True
print(is_true)
is_false = False
print(is_false)
```

special literals are used to represent special values in Python. The most common special literal is None, which represents the absence of a value or a null value.

```python
none_literal = None
print(none_literal)
```
