Gusessing Game 

```python
from random import randint 

random_number = randint(1, 100)
jackpot = random_number

guess_number = int(input("Enter a number between 1 and 100: "))
count = 1
while guess_number != jackpot:
    if guess_number < jackpot:
        print("Your guess is too low. Try again ->", guess_number)
    else:
        print("Your guess is too high. Try again ->", guess_number)
    guess_number = int(input("Enter a number between 1 and 100: "))
    count += 1
print("Congratulations! You guessed the correct number:", jackpot)
print("It took you", count, "attempts to guess the number.")
```
