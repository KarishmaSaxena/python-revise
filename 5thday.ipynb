TYPE CASTING refers to the conversion of one data type into another data type. In Python, we can perform type casting using built-in functions like int(), float(), str(), list(), tuple(), set(), dict(), etc.

```python
type_conversion = int(10.82)
print(type_conversion)
type_conversion = float(10)
print(type_conversion)
type_conversion = complex(10)
print(type_conversion)
type_conversion = str(10)
print(type_conversion)
type_conversion = bool(10)
print(type_conversion)
type_conversion = list((1,2,3))
print(type_conversion)
type_conversion = list('Karishma')
print(type_conversion)
TYPE_CONVERSION = str(10.82)
print(TYPE_CONVERSION)
```
